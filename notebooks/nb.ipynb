```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.model_selection import train_test_split
import catboost
from sklearn.metrics import root_mean_squared_error
```

```python
data = pd.read_csv('../data/train.csv')
data = data.dropna()
X = data.drop(['index', 'IC50, mM', 'CC50, mM', 'SI'], axis=1)

y_0 = pd.DataFrame(data['IC50, mM'])
y_1 = pd.DataFrame(data['CC50, mM'])
```

```python
def predict (X):
    model = catboost.CatBoostRegressor(random_seed=42)
    Xt, Xp, yt, yp = train_test_split(X, y_0, random_state=42)
    model.fit(Xt, yt)
    preds = model.predict(Xp)
    err = root_mean_squared_error(preds, yp)
    return err
```

```python
with open('./results.txt', 'w') as f:
    f.write('')
for k in range(1,200):
    selector = SelectKBest(score_func=f_regression, k=k)
    x_new = selector.fit_transform(X, y_0)
    indexes = [int(i) for i in selector.get_support(indices=1)]
    kX = X[X.columns[indexes]]
    result = predict(kX)
    with open('./results.txt', 'a') as f:
        f.write(f'{len(kX.columns)} {result}\n')
```
